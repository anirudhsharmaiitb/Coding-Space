## Multiplication Problem

We have to multiply two very large numbers efficiently. We will be using two algorithms - 1. Third Grade Multiplication, 2. Karatsuba Multiplication

We will also compare performance within the two. And how they stack up against python default.

Number 1 - 3141592653589793238462643383279502884197169399375105820974944592 <br>
Number 2 - 2718281828459045235360287471352662497757247093699959574966967627

```python
from time import time
```

```python
### Third Grade Multiplication

def third_grade_multiplication(num1, num2):
    len1 = len(num1)
    len2 = len(num2)

    num1_reverse = num1[::-1]
    num2_reverse = num2[::-1]

    first_product_list = []
    for i in range(len2):
        n = int(num2_reverse[i:i+1])
        product_str = ''
        carry = '0'
        for j in range(len1):
            m = int(num1_reverse[j:j+1])
            product = n*m + int(carry)
            product_str = product_str + str(product)[-1]
            carry = str(product)[:-1] if len(str(product)) > 1 else '0'
        first_product_list.append(carry+product_str[::-1])

    final_product = ''
    for i in range(len(first_product_list)-1):
        num1 = first_product_list[i] if i == 0 else final_product
        num2 = first_product_list[i+1]+'0'*(i+1)

        len1 = len(num1)
        len2 = len(num2)

        max_len = max(len1, len2)

        carry = '0'
        final_product = ''
        for j in range(max_len):
            n = int(num1[len1-j-1:len1-j]) if j < len1 else 0
            m = int(num2[len2-j-1:len2-j]) if j < len2 else 0
            product = n + m + int(carry)
            final_product = str(product)[-1] + final_product
            carry = str(product)[:-1] if len(str(product)) > 1 else '0'

    return final_product
```

```python
### Karatsuba Multiplication

def karatsuba(x, y):
    if x < 10 or y < 10:
        return x * y

    n = max(len(str(x)), len(str(y)))
    half_n = n // 2

    a = x // 10**half_n
    b = x % 10**half_n
    c = y // 10**half_n
    d = y % 10**half_n

    ac = karatsuba(a, c)
    bd = karatsuba(b, d)
    abcd = karatsuba(a + b, c + d)

    return (ac * 10**(2 * half_n)) + ((abcd - ac - bd) * 10**half_n) + bd
```

```python
def python_multiplication(x, y):
    return x * y
```

```python
num1 = '3141592653589793238462643383279502884197169399375105820974944592'
num2 = '2718281828459045235360287471352662497757247093699959574966967627'
```

```python
start_time = time()
third_grade_result = third_grade_multiplication(num1, num2)
end_time = time()
print(f"Third Grade Multiplication Time: {end_time - start_time:.6f} seconds")
print(third_grade_result)
```

```python
start_time = time()
karatsuba_result = karatsuba(int(num1), int(num2))
end_time = time()
print(f"Karastuba Multiplication Time: {end_time - start_time:.6f} seconds")
print(karatsuba_result)
```

```python
start_time = time()
python_multiplication_result = python_multiplication(int(num1), int(num2))
end_time = time()
print(f"Python Multiplication Time: {end_time - start_time:.6f} seconds")
print(python_multiplication_result)
```

#### Results

Clearly Karastuba was faster than Third Grade Multiplication by an order of 5x
